# Operadora simulada

Envie um lote sintético, consulte o demonstrativo e interprete as decisões sem acesso à rede.

```python
from tiss import examples, parse_demonstrativo, validate
from tiss.sandbox import demonstrativo, enviar_lote

lote = examples.load("lote-valido")
resposta = enviar_lote(lote)
assert resposta["aceito"]
protocolo = resposta["protocolo"]
resposta["numero_protocolo"]
```

```python
xml = demonstrativo(protocolo)
assert validate(xml).valid
decisoes = parse_demonstrativo(xml)
decisoes
```

O cenário `sem_autorizacao` glosa todos os itens da primeira guia e mantém as demais pagas.

```python
resposta_com_glosa = enviar_lote(lote, cenario="sem_autorizacao")
xml_com_glosa = demonstrativo(resposta_com_glosa["protocolo"])
assert validate(xml_com_glosa).valid
decisoes_com_glosa = parse_demonstrativo(xml_com_glosa)
decisoes_com_glosa
```

Na API hospedada, as rotas `/v1/sandbox/lotes` respondem sem chave.
