# Demonstrativo e conciliação

O demonstrativo de exemplo é lido sem enviar dados a um serviço. A decisão pode ser comparada com guias da aplicação.

```python
from voa_tiss import examples, parse_demonstrativo

decisoes = parse_demonstrativo(examples.load("demonstrativo"))
[(item.guia_ref, item.decision) for item in decisoes]
```

A conciliação propõe ações sem alterar os objetos recebidos. Um identificador estável permite aplicar a proposta na aplicação chamadora.

```python
from voa_tiss import plan_reconciliation

guias = [{"id": "guia-1", "numero_guia_prestador": decisoes[0].guia_ref, "status": "em_analise"}]
resultado = plan_reconciliation(decisoes, guias)
[(item.acao, item.guia["id"]) for item in resultado.propostas]
```
